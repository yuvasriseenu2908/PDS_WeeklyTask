```python
# Number of rows before removing null values
print("Rows before removing null values:", df.shape[0])

# Remove rows containing null values
df = df.dropna()

# Number of rows after b removing null values
print("Rows after removing null values:", df.shape[0])
```

```python
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
print("Final shape:", df.shape)

print("\nTotal Missing Values:", df.isnull().sum().sum())

print("Total Duplicate Rows:", df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
# Define the numerical columns present in your dataframe
numerical_features = [
    'wheel-base', 'length', 'width', 'curb-weight', 
    'engine-size', 'compression-ratio', 'city-mpg', 'highway-mpg'
]

summary = pd.DataFrame({
    "Mean": df[numerical_features].mean(),
    "Median": df[numerical_features].median(),
    "Minimum": df[numerical_features].min(),
    "Maximum": df[numerical_features].max(),
    "Standard Deviation": df[numerical_features].std()
})

summary
```

```python
# Category-wise car count
category = df.groupby("body-style")["make"].count()

category.plot(kind="bar", figsize=(7,4))
plt.title("Category-wise Car Count")
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()
```

```python
# Drive wheels vs average price
df["price"] = pd.to_numeric(df["price"], errors="coerce")
region = df.groupby("drive-wheels")["price"].mean()

region.plot(kind="bar", figsize=(7,4))
plt.title("Drive Wheels vs Average Price")
plt.xlabel("Drive Wheels")
plt.ylabel("Average Price")
plt.show()
```

```python
# 1. Scatter Plot
plt.figure(figsize=(6,4))
sns.scatterplot(data=df, x="horsepower", y="price")
plt.title("Horsepower vs Price")
plt.show()
```

```python
# 2. Bar Plot
plt.figure(figsize=(10,5))
df["make"].value_counts().plot(kind="bar")
plt.show()
plt.title("Car Brands")
plt.xlabel("Brand")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.show()
```

```python
# 3. Histogram
plt.figure(figsize=(6,4))
plt.hist(df["price"].dropna(), bins=15)
plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

```python
# 4. Box Plot
plt.figure(figsize=(6,4))
sns.boxplot(y=df["price"])
plt.title("Price Box Plot")
plt.show()
```

```python
# 6. Line Plot
plt.figure(figsize=(8,4))
sns.lineplot(x=df.index, y=df["price"])
plt.title("Car Price Line Plot")
plt.xlabel("Car Index")
plt.ylabel("Price")
plt.show()
```

```python

# 5. Correlation Heatmap
plt.figure(figsize=(12,8))
sns.heatmap(
    df.select_dtypes(include="number").corr(),
    annot=True, cmap="coolwarm", fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.show()
```
