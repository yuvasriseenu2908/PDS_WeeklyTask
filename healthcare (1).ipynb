```python
# ==========================================
# HEALTHCARE DATASET - COMPLETE ANALYSIS
# ==========================================

# 1. AGE GROUP ANALYSIS
df["Age Group"] = pd.cut(
    df["Age"],
    bins=[0, 18, 30, 45, 60, 75, 100],
    labels=["0-18", "19-30", "31-45", "46-60", "61-75", "76+"]
)

print("Age Group Distribution:")
print(df["Age Group"].value_counts().sort_index())

print("\nAge Group by Medical Condition:")
print(pd.crosstab(df["Medical Condition"], df["Age Group"]))


# 2. EMERGENCY ADMISSIONS BY MEDICAL CONDITION
emergency = df[df["Admission Type"] == "Emergency"]

print("\nEmergency Admissions by Medical Condition:")
print(emergency["Medical Condition"].value_counts())


# 3. ADMISSION TYPE BY MEDICAL CONDITION
print("\nAdmission Type by Medical Condition:")
print(pd.crosstab(df["Medical Condition"], df["Admission Type"]))


# 4. AVERAGE BILLING BY INSURANCE PROVIDER
print("\nAverage Billing by Insurance Provider:")
print(
    df.groupby("Insurance Provider")["Billing Amount"]
      .mean()
      .sort_values(ascending=False)
)


# 5. BILLING BY ADMISSION TYPE
print("\nBilling by Admission Type:")
print(
    df.groupby("Admission Type")["Billing Amount"]
      .agg(["mean", "median", "std", "min", "max"])
)


# 6. LENGTH OF STAY
df["Length of Stay"] = (
    df["Discharge Date"] - df["Date of Admission"]
).dt.days

print("\nAverage Length of Stay:")
print(df["Length of Stay"].mean())


# 7. LENGTH OF STAY BY MEDICAL CONDITION
print("\nAverage Length of Stay by Medical Condition:")
print(
    df.groupby("Medical Condition")["Length of Stay"]
      .mean()
      .sort_values(ascending=False)
)


# 8. BILLING BY LENGTH OF STAY
print("\nBilling and Length of Stay:")
print(
    df[["Length of Stay", "Billing Amount"]].describe()
)


# 9. MONTHLY ADMISSIONS
df["Admission Month"] = df["Date of Admission"].dt.month

monthly_admissions = df["Admission Month"].value_counts().sort_index()

print("\nMonthly Admissions:")
print(monthly_admissions)

print("\nMonth with Highest Admissions:")
print(monthly_admissions.idxmax())

print("\nMonth with Lowest Admissions:")
print(monthly_admissions.idxmin())


# 10. MONTHLY ADMISSIONS WITH MONTH NAMES
df["Month Name"] = df["Date of Admission"].dt.month_name()

monthly = (
    df.groupby(df["Date of Admission"].dt.month)
      .size()
      .sort_values(ascending=False)
)

print("\nAdmissions by Month:")
print(monthly)


# 11. CORRELATION ANALYSIS
numerical_columns = [
    "Age",
    "Billing Amount",
    "Length of Stay",
    "Room Number"
]

correlation = df[numerical_columns].corr()

print("\nCorrelation Matrix:")
print(correlation)


# 12. STRONGEST CORRELATIONS
corr_pairs = (
    correlation.where(
        np.triu(np.ones(correlation.shape), k=1).astype(bool)
    )
    .stack()
    .sort_values(key=abs, ascending=False)
)

print("\nStrongest Correlations:")
print(corr_pairs)
```
