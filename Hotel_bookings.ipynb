```python
df = df.drop_duplicates()
```

```python
df['children'] = df['children'].fillna(0)
df['country'] = df['country'].fillna('Unknown')
df['agent'] = df['agent'].fillna(0)
df['company'] = df['company'].fillna(0)
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['lead_time'], bins=30)
plt.title("Distribution of Lead Time")
plt.xlabel("Lead Time")
plt.ylabel("Frequency")
plt.show()
```

```python

plt.figure(figsize=(7,5))
sns.countplot(data=df, x='hotel')
plt.title("Hotel Booking Count")
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.countplot(data=df, x='market_segment')
plt.title("Bookings by Market Segment")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['lead_time'], df['adr'], alpha=0.3)
plt.title("Lead Time vs ADR")
plt.xlabel("Lead Time")
plt.ylabel("ADR")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='hotel', y='adr')
plt.title("ADR by Hotel Type")
plt.show()
```

```python
yearly = df.groupby('arrival_date_year').size()

plt.figure(figsize=(8,5))
yearly.plot(kind='line', marker='o')
plt.title("Yearly Hotel Bookings")
plt.xlabel("Year")
plt.ylabel("Number of Bookings")
plt.grid()
plt.show()
```

```python
monthly = df.groupby('arrival_date_month').size()

month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

monthly = monthly.reindex(month_order)

plt.figure(figsize=(10,5))
monthly.plot(kind='line', marker='o')
plt.title("Monthly Hotel Bookings")
plt.xlabel("Month")
plt.ylabel("Bookings")
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.countplot(data=df, x='hotel', hue='is_canceled')
plt.title("Cancellation by Hotel")
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.countplot(data=df, x='customer_type')
plt.title("Bookings by Customer Type")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='deposit_type')
plt.title("Bookings by Deposit Type")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='total_of_special_requests')
plt.title("Total Special Requests")
plt.show()
```

```python
top_countries = df['country'].value_counts().head(10)

plt.figure(figsize=(10,5))
top_countries.plot(kind='bar')
plt.title("Top 10 Countries by Bookings")
plt.xlabel("Country")
plt.ylabel("Bookings")
plt.xticks(rotation=45)
plt.show()
```

```python
adr = df.groupby('hotel')['adr'].mean()

plt.figure(figsize=(7,5))
adr.plot(kind='bar')
plt.title("Average ADR by Hotel")
plt.ylabel("Average ADR")
plt.show()
```
