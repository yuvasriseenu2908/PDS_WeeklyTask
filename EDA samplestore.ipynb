```python
# ============================================================
# EDA - SAMPLE SUPERSTORE
# SALES, PROFIT, DISCOUNT & CORRELATION INSIGHTS
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# 1. LOAD DATA
# ------------------------------------------------------------

df = pd.read_csv(
    r"C:\Users\Yuvasri\Desktop\Weekly Task\samplesuperstore.csv"
)

# Remove missing values
df = df.dropna()

print("=" * 70)
print("              SAMPLE SUPERSTORE EDA")
print("=" * 70)

print("\nDataset Shape:", df.shape)


# ============================================================
# 2. SALES INSIGHTS
# ============================================================

print("\n\n" + "=" * 70)
print("                       SALES INSIGHTS")
print("=" * 70)

# Sales by Category
category_sales = (
    df.groupby("Category")["Sales"]
      .sum()
      .sort_values(ascending=False)
)

print("\n1. Sales by Category:")
display(category_sales)

print("\nHighest Sales Category:")
print(category_sales.idxmax(), "=", round(category_sales.max(), 2))

print("\nLowest Sales Category:")
print(category_sales.idxmin(), "=", round(category_sales.min(), 2))


# Sales by Region
region_sales = (
    df.groupby("Region")["Sales"]
      .sum()
      .sort_values(ascending=False)
)

print("\n2. Sales by Region:")
display(region_sales)

print("\nRegion contributing most to Sales:")
print(region_sales.idxmax(), "=", round(region_sales.max(), 2))


# ============================================================
# 3. PROFIT INSIGHTS
# ============================================================

print("\n\n" + "=" * 70)
print("                       PROFIT INSIGHTS")
print("=" * 70)

# Profit by Category
category_profit = (
    df.groupby("Category")["Profit"]
      .sum()
      .sort_values(ascending=False)
)

print("\n1. Profit by Category:")
display(category_profit)

print("\nCategory generating highest profit:")
print(category_profit.idxmax(), "=", round(category_profit.max(), 2))


# Profit by Sub-Category
subcategory_profit = (
    df.groupby("Sub-Category")["Profit"]
      .sum()
      .sort_values()
)

print("\n2. Profit by Sub-Category:")
display(subcategory_profit)

print("\nSub-Categories with negative profit:")
negative_subcategories = subcategory_profit[
    subcategory_profit < 0
]

display(negative_subcategories)


# Profit by Region
region_profit = (
    df.groupby("Region")["Profit"]
      .sum()
      .sort_values(ascending=False)
)

print("\n3. Profit by Region:")
display(region_profit)

print("\nMost profitable region:")
print(region_profit.idxmax(), "=", round(region_profit.max(), 2))

print("\nLeast profitable region:")
print(region_profit.idxmin(), "=", round(region_profit.min(), 2))


# ============================================================
# 4. DISCOUNT INSIGHTS
# ============================================================

print("\n\n" + "=" * 70)
print("                     DISCOUNT INSIGHTS")
print("=" * 70)

# Profit by Discount
discount_profit = (
    df.groupby("Discount")["Profit"]
      .agg(["mean", "sum", "count"])
      .sort_index()
)

print("\n1. Profit by Discount Level:")
display(discount_profit)


# Discount-Profit Correlation
discount_profit_corr = df["Discount"].corr(df["Profit"])

print("\n2. Discount vs Profit Correlation:")
print(round(discount_profit_corr, 3))

if discount_profit_corr < 0:
    print("Conclusion: Profit tends to decrease as Discount increases.")
else:
    print("Conclusion: Discount does not have a negative relationship with Profit.")


# Poor profitability discount levels
poor_discount_levels = discount_profit[
    discount_profit["mean"] < 0
]

print("\n3. Discount levels with poor profitability:")
display(poor_discount_levels)


# High discount + negative profit
high_discount_negative_profit = df[
    (df["Discount"] >= 0.50) &
    (df["Profit"] < 0)
]

print("\n4. High Discount (>= 50%) with Negative Profit:")
display(
    high_discount_negative_profit[
        [
            "Order ID",
            "Product Name",
            "Category",
            "Sub-Category",
            "Sales",
            "Discount",
            "Profit"
        ]
    ].sort_values("Discount", ascending=False)
)

print(
    "\nNumber of high-discount negative-profit transactions:",
    len(high_discount_negative_profit)
)


# ============================================================
# 5. CORRELATION INSIGHTS
# ============================================================

print("\n\n" + "=" * 70)
print("                    CORRELATION INSIGHTS")
print("=" * 70)

correlation_columns = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]

correlation_matrix = df[correlation_columns].corr()

print("\n1. Correlation Matrix:")
display(correlation_matrix)


# Positive correlations
positive_correlations = []

for i in range(len(correlation_columns)):
    for j in range(i + 1, len(correlation_columns)):

        var1 = correlation_columns[i]
        var2 = correlation_columns[j]
        value = correlation_matrix.loc[var1, var2]

        if value > 0:
            positive_correlations.append(
                [var1, var2, value]
            )

positive_df = pd.DataFrame(
    positive_correlations,
    columns=["Variable 1", "Variable 2", "Correlation"]
).sort_values(
    "Correlation",
    ascending=False
)

print("\n2. Positive Correlations:")
display(positive_df)


# Negative correlations
negative_correlations = []

for i in range(len(correlation_columns)):
    for j in range(i + 1, len(correlation_columns)):

        var1 = correlation_columns[i]
        var2 = correlation_columns[j]
        value = correlation_matrix.loc[var1, var2]

        if value < 0:
            negative_correlations.append(
                [var1, var2, value]
            )

negative_df = pd.DataFrame(
    negative_correlations,
    columns=["Variable 1", "Variable 2", "Correlation"]
).sort_values(
    "Correlation"
)

print("\n3. Negative Correlations:")
display(negative_df)


print("\n4. Is Discount negatively related to Profit?")

if discount_profit_corr < 0:
    print("YES - Discount is negatively related to Profit.")
else:
    print("NO - Discount is not negatively related to Profit.")


# ============================================================
# 6. VISUALIZATIONS
# ============================================================

print("\n\n" + "=" * 70)
print("                       VISUALIZATIONS")
print("=" * 70)


# ---------------- SALES BY CATEGORY ----------------

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Category",
    y="Sales",
    estimator="sum"
)

plt.title("Category-wise Total Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()


# ---------------- SALES BY REGION ----------------

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Region",
    y="Sales",
    estimator="sum"
)

plt.title("Region-wise Total Sales")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.show()


# ---------------- PROFIT BY CATEGORY ----------------

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Category",
    y="Profit",
    estimator="sum"
)

plt.title("Category-wise Total Profit")
plt.xlabel("Category")
plt.ylabel("Total Profit")

plt.show()


# ---------------- PROFIT BY SUB-CATEGORY ----------------

plt.figure(figsize=(10, 6))

sns.barplot(
    x=subcategory_profit.values,
    y=subcategory_profit.index
)

plt.title("Sub-Category-wise Total Profit")
plt.xlabel("Total Profit")
plt.ylabel("Sub-Category")

plt.show()


# ---------------- PROFIT BY REGION ----------------

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Region",
    y="Profit",
    estimator="sum"
)

plt.title("Region-wise Total Profit")
plt.xlabel("Region")
plt.ylabel("Total Profit")

plt.show()


# ---------------- DISCOUNT VS PROFIT ----------------

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit"
)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.show()


# ---------------- CORRELATION HEATMAP ----------------

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()


# ============================================================
# 7. FINAL KEY FINDINGS
# ============================================================

print("\n\n" + "=" * 70)
print("                       KEY FINDINGS")
print("=" * 70)

print("\n1. Highest Sales Category       :", category_sales.idxmax())
print("2. Lowest Sales Category        :", category_sales.idxmin())
print("3. Top Sales Region             :", region_sales.idxmax())
print("4. Highest Profit Category      :", category_profit.idxmax())
print("5. Most Profitable Region       :", region_profit.idxmax())
print("6. Least Profitable Region      :", region_profit.idxmin())
print("7. Discount-Profit Correlation  :", round(discount_profit_corr, 3))
print("8. High Discount Negative Profit Transactions:",
      len(high_discount_negative_profit))

print("\n" + "=" * 70)
print("                    EDA COMPLETED")
print("=" * 70)
```
