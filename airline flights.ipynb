```python
# ============================================
# BIVARIATE ANALYSIS
# ============================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("airlines_flights_data (5).csv")

# Display columns
print("Columns in Dataset:")
print(df.columns.tolist())


# ------------------------------------------------
# 1. Airline vs Price
# ------------------------------------------------

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x='airline',
    y='price'
)

plt.title("Flight Price by Airline")
plt.xlabel("Airline")
plt.ylabel("Price")
plt.xticks(rotation=45)
plt.show()


# ------------------------------------------------
# 2. Class vs Price
# ------------------------------------------------

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x='class',
    y='price'
)

plt.title("Flight Price by Class")
plt.xlabel("Class")
plt.ylabel("Price")
plt.show()


# ------------------------------------------------
# 3. Stops vs Price
# ------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x='stops',
    y='price'
)

plt.title("Flight Price by Number of Stops")
plt.xlabel("Number of Stops")
plt.ylabel("Price")
plt.show()


# ------------------------------------------------
# 4. Source City vs Price
# ------------------------------------------------

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x='source_city',
    y='price'
)

plt.title("Flight Price by Source City")
plt.xlabel("Source City")
plt.ylabel("Price")
plt.show()


# ------------------------------------------------
# 5. Destination City vs Price
# ------------------------------------------------

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x='destination_city',
    y='price'
)

plt.title("Flight Price by Destination City")
plt.xlabel("Destination City")
plt.ylabel("Price")
plt.show()


# ------------------------------------------------
# 6. Departure Time vs Price
# ------------------------------------------------

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x='departure_time',
    y='price'
)

plt.title("Flight Price by Departure Time")
plt.xlabel("Departure Time")
plt.ylabel("Price")
plt.xticks(rotation=45)
plt.show()


# ------------------------------------------------
# 7. Arrival Time vs Price
# ------------------------------------------------

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x='arrival_time',
    y='price'
)

plt.title("Flight Price by Arrival Time")
plt.xlabel("Arrival Time")
plt.ylabel("Price")
plt.xticks(rotation=45)
plt.show()


# ------------------------------------------------
# 8. Days Left vs Price
# ------------------------------------------------

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='days_left',
    y='price',
    alpha=0.4
)

plt.title("Days Left vs Flight Price")
plt.xlabel("Days Left")
plt.ylabel("Price")
plt.show()


# ------------------------------------------------
# 9. Duration vs Price
# ------------------------------------------------

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='duration',
    y='price',
    alpha=0.4
)

plt.title("Flight Duration vs Price")
plt.xlabel("Duration (Hours)")
plt.ylabel("Price")
plt.show()


# ------------------------------------------------
# 10. Airline vs Class vs Price
# ------------------------------------------------

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x='airline',
    y='price',
    hue='class'
)

plt.title("Flight Price by Airline and Class")
plt.xlabel("Airline")
plt.ylabel("Price")
plt.xticks(rotation=45)
plt.show()


# ------------------------------------------------
# 11. Airline vs Stops
# ------------------------------------------------

plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    x='airline',
    hue='stops'
)

plt.title("Number of Stops by Airline")
plt.xlabel("Airline")
plt.ylabel("Number of Flights")
plt.xticks(rotation=45)
plt.show()


# ------------------------------------------------
# 12. Source City vs Destination City
# ------------------------------------------------

route_count = pd.crosstab(
    df['source_city'],
    df['destination_city']
)

print("\nSource City vs Destination City:")
print(route_count)

plt.figure(figsize=(10, 7))

sns.heatmap(
    route_count,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.title("Flight Count: Source City vs Destination City")
plt.xlabel("Destination City")
plt.ylabel("Source City")
plt.show()


# ------------------------------------------------
# 13. Departure Time vs Class
# ------------------------------------------------

plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    x='departure_time',
    hue='class'
)

plt.title("Departure Time by Class")
plt.xlabel("Departure Time")
plt.ylabel("Number of Flights")
plt.xticks(rotation=45)
plt.show()


# ------------------------------------------------
# 14. Correlation between Numerical Variables
# ------------------------------------------------

numeric_columns = [
    'duration',
    'days_left',
    'price'
]

correlation = df[numeric_columns].corr()

print("\nCorrelation Matrix:")
print(correlation)

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")
plt.show()
```

# TASK 4: BUSINESS INSIGHTS

```python
# ============================================
# BUSINESS INSIGHTS
# ============================================

import pandas as pd

# Load dataset
df = pd.read_csv("airlines_flights_data (5).csv")


# ------------------------------------------------
# 1. Most Popular Airline
# ------------------------------------------------

airline_counts = df['airline'].value_counts()

most_popular_airline = airline_counts.idxmax()
most_popular_airline_count = airline_counts.max()

print("1. MOST POPULAR AIRLINE")
print("------------------------")
print("Airline:", most_popular_airline)
print("Number of Flights:", most_popular_airline_count)


# ------------------------------------------------
# 2. Least Popular Airline
# ------------------------------------------------

least_popular_airline = airline_counts.idxmin()
least_popular_airline_count = airline_counts.min()

print("\n2. LEAST POPULAR AIRLINE")
print("-------------------------")
print("Airline:", least_popular_airline)
print("Number of Flights:", least_popular_airline_count)


# ------------------------------------------------
# 3. Average Price by Airline
# ------------------------------------------------

average_airline_price = (
    df.groupby('airline')['price']
    .mean()
    .sort_values(ascending=False)
)

print("\n3. AVERAGE PRICE BY AIRLINE")
print("----------------------------")
print(average_airline_price)


# ------------------------------------------------
# 4. Most Expensive Airline
# ------------------------------------------------

most_expensive_airline = average_airline_price.idxmax()
highest_average_price = average_airline_price.max()

print("\n4. AIRLINE WITH HIGHEST AVERAGE PRICE")
print("--------------------------------------")
print("Airline:", most_expensive_airline)
print("Average Price:", round(highest_average_price, 2))


# ------------------------------------------------
# 5. Lowest Average Price Airline
# ------------------------------------------------

cheapest_airline = average_airline_price.idxmin()
lowest_average_price = average_airline_price.min()

print("\n5. AIRLINE WITH LOWEST AVERAGE PRICE")
print("-------------------------------------")
print("Airline:", cheapest_airline)
print("Average Price:", round(lowest_average_price, 2))


# ------------------------------------------------
# 6. Average Price by Class
# ------------------------------------------------

class_price = (
    df.groupby('class')['price']
    .mean()
    .sort_values(ascending=False)
)

print("\n6. AVERAGE PRICE BY CLASS")
print("-------------------------")
print(class_price)


# ------------------------------------------------
# 7. Average Price by Stops
# ------------------------------------------------

stops_price = (
    df.groupby('stops')['price']
    .mean()
    .sort_values(ascending=False)
)

print("\n7. AVERAGE PRICE BY STOPS")
print("-------------------------")
print(stops_price)


# ------------------------------------------------
# 8. Most Common Stop Type
# ------------------------------------------------

stop_counts = df['stops'].value_counts()

most_common_stops = stop_counts.idxmax()
most_common_stops_count = stop_counts.max()

print("\n8. MOST COMMON STOP TYPE")
print("------------------------")
print("Stops:", most_common_stops)
print("Number of Flights:", most_common_stops_count)


# ------------------------------------------------
# 9. Most Common Source City
# ------------------------------------------------

source_counts = df['source_city'].value_counts()

most_common_source = source_counts.idxmax()

print("\n9. MOST COMMON SOURCE CITY")
print("--------------------------")
print("City:", most_common_source)
print("Number of Flights:", source_counts.max())


# ------------------------------------------------
# 10. Most Common Destination City
# ------------------------------------------------

destination_counts = df['destination_city'].value_counts()

most_common_destination = destination_counts.idxmax()

print("\n10. MOST COMMON DESTINATION CITY")
print("--------------------------------")
print("City:", most_common_destination)
print("Number of Flights:", destination_counts.max())


# ------------------------------------------------
# 11. Most Common Route
# ------------------------------------------------

df['route'] = (
    df['source_city'] + " -> " + df['destination_city']
)

route_counts = df['route'].value_counts()

most_common_route = route_counts.idxmax()

print("\n11. MOST COMMON ROUTE")
print("---------------------")
print("Route:", most_common_route)
print("Number of Flights:", route_counts.max())


# ------------------------------------------------
# 12. Average Flight Duration
# ------------------------------------------------

average_duration = df['duration'].mean()

print("\n12. AVERAGE FLIGHT DURATION")
print("---------------------------")
print("Average Duration:",
      round(average_duration, 2),
      "hours")


# ------------------------------------------------
# 13. Average Flight Price
# ------------------------------------------------

average_price = df['price'].mean()

print("\n13. OVERALL AVERAGE FLIGHT PRICE")
print("--------------------------------")
print("Average Price:", round(average_price, 2))


# ------------------------------------------------
# 14. Cheapest Flight
# ------------------------------------------------

cheapest_flight = df.loc[df['price'].idxmin()]

print("\n14. CHEAPEST FLIGHT")
print("-------------------")
print(cheapest_flight)


# ------------------------------------------------
# 15. Most Expensive Flight
# ------------------------------------------------

expensive_flight = df.loc[df['price'].idxmax()]

print("\n15. MOST EXPENSIVE FLIGHT")
print("-------------------------")
print(expensive_flight)


# ------------------------------------------------
# 16. Average Price by Source City
# ------------------------------------------------

source_price = (
    df.groupby('source_city')['price']
    .mean()
    .sort_values(ascending=False)
)

print("\n16. AVERAGE PRICE BY SOURCE CITY")
print("--------------------------------")
print(source_price)


# ------------------------------------------------
# 17. Average Price by Destination City
# ------------------------------------------------

destination_price = (
    df.groupby('destination_city')['price']
    .mean()
    .sort_values(ascending=False)
)

print("\n17. AVERAGE PRICE BY DESTINATION CITY")
print("-------------------------------------")
print(destination_price)


# ------------------------------------------------
# 18. Average Price by Departure Time
# ------------------------------------------------

departure_price = (
    df.groupby('departure_time')['price']
    .mean()
    .sort_values(ascending=False)
)

print("\n18. AVERAGE PRICE BY DEPARTURE TIME")
print("-----------------------------------")
print(departure_price)


# ------------------------------------------------
# 19. Average Price by Arrival Time
# ------------------------------------------------

arrival_price = (
    df.groupby('arrival_time')['price']
    .mean()
    .sort_values(ascending=False)
)

print("\n19. AVERAGE PRICE BY ARRIVAL TIME")
print("---------------------------------")
print(arrival_price)


# ------------------------------------------------
# 20. Average Price by Days Left
# ------------------------------------------------

days_price = (
    df.groupby('days_left')['price']
    .mean()
)

print("\n20. AVERAGE PRICE BY DAYS LEFT")
print("------------------------------")
print(days_price)


# ------------------------------------------------
# 21. Correlation Analysis
# ------------------------------------------------

correlation = df[
    ['duration', 'days_left', 'price']
].corr()

print("\n21. CORRELATION MATRIX")
print("---------------------")
print(correlation)


# ------------------------------------------------
# 22. Overall Business Summary
# ------------------------------------------------

print("\n")
print("==============================================")
print("           BUSINESS INSIGHTS SUMMARY")
print("==============================================")

print(
    "Most Popular Airline:",
    most_popular_airline
)

print(
    "Most Common Route:",
    most_common_route
)

print(
    "Most Common Source City:",
    most_common_source
)

print(
    "Most Common Destination City:",
    most_common_destination
)

print(
    "Most Common Stop Type:",
    most_common_stops
)

print(
    "Average Flight Price:",
    round(average_price, 2)
)

print(
    "Average Flight Duration:",
    round(average_duration, 2),
    "hours"
)

print(
    "Highest Average Price Airline:",
    most_expensive_airline
)

print(
    "Lowest Average Price Airline:",
    cheapest_airline
)

print("==============================================")
```

```python
# ============================================
# CORRELATION HEATMAP - AIRLINE FLIGHTS DATA
# ============================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("airlines_flights_data (5).csv")

# Select numerical columns
numeric_data = df.select_dtypes(include=['int64', 'float64'])

# Calculate correlation
correlation = numeric_data.corr()

# Display correlation values
print("Correlation Matrix:")
print(correlation)

# Create heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title("Correlation Heatmap - Airlines Flights Dataset")
plt.show()
```
