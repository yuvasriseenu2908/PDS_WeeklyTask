```python
df = df.drop_duplicates()
```

```python
df['date'] = pd.to_datetime(df['date'], utc=True)
```

```python
print(df.dtypes)
```

```python
df = df.dropna()
```

```python
print("Average Close:", df['close'].mean())
```

```python
print("Maximum Close:", df['close'].max())
```

```python
print("Minimum Close:", df['close'].min())
```

```python
print("Total Volume:", df['volume'].sum())
```

```python
df['year'] = df['date'].dt.year
```

```python
df['month'] = df['date'].dt.month
```

```python
# Histogram
plt.figure(figsize=(8,5))
plt.hist(df['close'], bins=30)
plt.title("Closing Price Distribution")
plt.xlabel("Close Price")
plt.ylabel("Frequency")
plt.show()
```

```python
#  Bar Chart
plt.figure(figsize=(8,5))
plt.bar(df['year'], df['volume'])
plt.title("Trading Volume by Year")
plt.xlabel("Year")
plt.ylabel("Volume")
plt.show()
```

```python
#  Scatter Plot
plt.figure(figsize=(8,5))
plt.scatter(df['open'], df['close'])
plt.title("Open vs Close Price")
plt.xlabel("Open Price")
plt.ylabel("Close Price")
plt.show()
```

```python
#  Box Plot
plt.figure(figsize=(8,5))
sns.boxplot(data=df[['open','high','low','close']])
plt.title("Stock Price Distribution")
plt.show()
```

```python
#  Line Plot - Closing Price
plt.figure(figsize=(12,5))
plt.plot(df['date'], df['close'])
plt.title("Stock Closing Price Over Time")
plt.xlabel("Date")
plt.ylabel("Close Price")
plt.show()
```

```python
#  Line Plot - Opening Price
plt.figure(figsize=(12,5))
plt.plot(df['date'], df['open'])
plt.title("Stock Opening Price Over Time")
plt.xlabel("Date")
plt.ylabel("Open Price")
plt.show()
```

```python
# . High and Low Price
plt.figure(figsize=(12,5))
plt.plot(df['date'], df['high'], label='High')
plt.plot(df['date'], df['low'], label='Low')
plt.title("High and Low Stock Prices")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.show()
```

```python

plt.figure(figsize=(12,5))
plt.plot(df['date'], df['adj_close'])
plt.title("Adjusted Closing Price")
plt.xlabel("Date")
plt.ylabel("Adjusted Close")
plt.show()
```

```python
#  Volume Distribution
plt.figure(figsize=(8,5))
plt.hist(df['volume'], bins=30)
plt.title("Trading Volume Distribution")
plt.xlabel("Volume")
plt.ylabel("Frequency")
plt.show()
```

```python
monthly = df.groupby('month')['close'].mean()

plt.figure(figsize=(8,5))
monthly.plot(kind='bar')
plt.title("Monthly Average Closing Price")
plt.xlabel("Month")
plt.ylabel("Average Close")
plt.show()
```

```python
yearly = df.groupby('year')['close'].mean()

plt.figure(figsize=(10,5))
yearly.plot(kind='bar')
plt.title("Yearly Average Closing Price")
plt.xlabel("Year")
plt.ylabel("Average Close")
plt.show()
```

```python
year_volume = df.groupby('year')['volume'].sum()

plt.figure(figsize=(10,5))
year_volume.plot(kind='bar')
plt.title("Yearly Trading Volume")
plt.xlabel("Year")
plt.ylabel("Total Volume")
plt.show()
```

```python
#  Correlation Heatmap
plt.figure(figsize=(8,6))
sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    annot=True,
    cmap='coolwarm'
)
plt.title("Correlation Heatmap")
plt.show()
```
